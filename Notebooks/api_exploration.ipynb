                 FPL API
                    │
       ┌────────────┼──────────────┐
       │            │              │
       ▼            ▼              ▼
bootstrap-static  fixtures    event/{gw}/live
       │            │              │
       │            │              │
       ▼            ▼              ▼
 master data     matches      player GW stats
       │            │              │
       │            │              ├── stats
       │            │              └── explain
       │            │
       └────────────┴──────────────┘

bootstrap-static = who/what exist and their current aggregate state
fixtures - what matches exist and what happened to them
event/hew/live - how did each player perform during the gameweek

| Endpoint                    | Main grain              | Purpose                              |
| --------------------------- | ----------------------- | ------------------------------------ |
| `bootstrap-static`          | Entity/master data      | Players, teams, positions, gameweeks |
| `fixtures`                  | Fixture                 | Matches, scores, match-level stats   |
| `event/{gw}/live`           | Player × Gameweek       | Player performance for a GW          |
| `event/{gw}/live → explain` | Player × Fixture × Stat | FPL point attribution                |



```python
print(len(gw1_live["elements"]))
print(gw1_live["elements"][-1])

#so in bootstrap static we had 623 total . gw1 live is showing 610 . and the last player hasn't even played 
#lets find out who are those 13 players missing 

master_player_ids = {
    player["id"]
    for player in data["elements"]
}

live_player_ids = {
    player["id"]
    for player in gw1_live["elements"]
}

missing_ids = master_player_ids - live_player_ids
print("Master players:", len(master_player_ids))
print("Live players:", len(live_player_ids))
print("Missing:", len(missing_ids))
print("Missing IDs:", sorted(missing_ids))
```

```python
for player in data["elements"]:
    if player["id"] in missing_ids:
        print(
            player["id"],
            player["web_name"],
            player["status"],
            player["team"],
            player["removed"]
        )

'''bootstrap-static is the broader/current player universe, while event/{gw}/live contains the players for whom the API provides gameweek-live records.'''
```

```python
for player in gw1_live["elements"]:
    if player["stats"]["total_points"] > 0:
        print("Player ID:", player["id"])
        print("Total points:", player["stats"]["total_points"])
        print("Explain:", player["explain"])
        break
```
